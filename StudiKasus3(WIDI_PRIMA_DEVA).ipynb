## Judul Studi Kasus: Bengkulu Snap & Solve: Purwarupa Web GIS untuk Pelaporan Aset dan Fasilitas Publik Menggunakan Leaflet.js dan Flask

```python
# 1. Install komponen Folium (Leaflet.js untuk Python)
!pip install folium pandas

import folium
import pandas as pd
from IPython.display import display
```

```python
# 2. Simulasi Data Laporan Aset
# Mengadopsi konsep "Snap Send Solve" dari pemaparan materi
data_laporan = {
    'id_laporan': [101, 102, 103],
    'jenis_kerusakan': [
        'Jalan Berlubang',
        'Pohon Rawan Tumbang',
        'Lampu Jalan Mati'
    ],
    'lat': [-3.7950, -3.8010, -3.7925],
    'lon': [102.2590, 102.2620, 102.2550],
    'status': [
        'Menunggu Tindakan',
        'Sedang Dikerjakan',
        'Selesai'
    ],
    'prioritas': ['Tinggi', 'Sedang', 'Rendah']
}

df_laporan = pd.DataFrame(data_laporan)

# Menampilkan data dummy
display(df_laporan)
```

```python
# 3. Inisialisasi Komponen Peta
peta_bengkulu = folium.Map(
    location=[-3.795, 102.259],
    zoom_start=14,
    tiles=None
)

folium.TileLayer(
    tiles='https://server.arcgisonline.com/ArcGIS/rest/services/World_Street_Map/MapServer/tile/{z}/{y}/{x}',
    attr='Tiles &copy; Esri',
    name='Esri World Street Map'
).add_to(peta_bengkulu)
```

```python
# 4. Melakukan iterasi data dan menambahkan Marker ke dalam Peta
for index, row in df_laporan.iterrows():
    # Mengatur warna marker berdasarkan prioritas perbaikan
    if row['prioritas'] == 'Tinggi':
        warna = 'red'
    elif row['prioritas'] == 'Sedang':
        warna = 'orange'
    else:
        warna = 'green'

    # Konten popup saat marker diklik
    popup_html = f"""
    <b>Kerusakan:</b> {row['jenis_kerusakan']}<br>
    <b>Status:</b> {row['status']}<br>
    <b>Prioritas:</b> {row['prioritas']}
    """

    # Menambahkan komponen Marker
    folium.Marker(
        location=[row['lat'], row['lon']],
        popup=folium.Popup(popup_html, max_width=250),
        tooltip="Klik untuk detail aset",
        icon=folium.Icon(color=warna, icon="info-sign")
    ).add_to(peta_bengkulu)
```

```python
#5. Menambahkan marker berdasarkan prioritas

for index, row in df_laporan.iterrows():

    # Menentukan warna berdasarkan prioritas
    if row['prioritas'] == 'Tinggi':
        warna = 'red'
    elif row['prioritas'] == 'Sedang':
        warna = 'orange'
    else:
        warna = 'green'

    # Membuat informasi popup
    popup_html = f"""
    <b>ID Laporan:</b> {row['id_laporan']}<br>
    <b>Kerusakan:</b> {row['jenis_kerusakan']}<br>
    <b>Status:</b> {row['status']}<br>
    <b>Prioritas:</b> {row['prioritas']}
    """

    # Menambahkan marker ke peta
    folium.Marker(
        location=[row['lat'], row['lon']],
        popup=folium.Popup(popup_html, max_width=300),
        tooltip=f"Laporan {row['id_laporan']}: {row['jenis_kerusakan']}",
        icon=folium.Icon(
            color=warna,
            icon='info-sign'
        )
    ).add_to(peta_bengkulu)
```

```python
# 5. Menampilkan peta interaktif
display(peta_bengkulu)
```
